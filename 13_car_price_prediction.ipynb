```python
# Shared imports (available in standard Google Colab)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_classification, make_regression, make_blobs, fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, mean_squared_error, r2_score

np.random.seed(42)
print("Setup complete")
```

```python
from sklearn.ensemble import RandomForestRegressor
rng=np.random.default_rng(22); n=500
df=pd.DataFrame({'age_years':rng.integers(0,16,n),'mileage_km':rng.integers(5000,220000,n),'engine_cc':rng.choice([1000,1200,1500,1800,2000,2500],n),'fuel_code':rng.integers(0,2,n)})
df['price']=np.maximum(1500,30000-df.age_years*1300-df.mileage_km*.055+df.engine_cc*3+df.fuel_code*1800+rng.normal(0,1800,n))
Xtr,Xte,ytr,yte=train_test_split(df.drop(columns='price'),df.price,test_size=.2,random_state=42)
m=RandomForestRegressor(n_estimators=150,random_state=42).fit(Xtr,ytr); p=m.predict(Xte)
print('RMSE:',round(mean_squared_error(yte,p)**.5,2),'R²:',round(r2_score(yte,p),3)); print('Example predicted price:',round(m.predict(pd.DataFrame([{'age_years':5,'mileage_km':60000,'engine_cc':1500,'fuel_code':1}]))[0],2))
```
