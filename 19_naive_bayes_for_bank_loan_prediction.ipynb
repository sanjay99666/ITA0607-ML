```python
# Shared imports (available in standard Google Colab)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_classification, make_regression, make_blobs, fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, mean_squared_error, r2_score

np.random.seed(42)
print("Setup complete")
```

```python
from sklearn.naive_bayes import GaussianNB
rng=np.random.default_rng(19); n=700
bank=pd.DataFrame({'income_k':rng.normal(65,22,n).clip(15),'loan_k':rng.uniform(2,90,n),'credit_score':rng.normal(680,65,n).clip(450,850),'employment_years':rng.integers(0,35,n)})
prob=1/(1+np.exp(-np.clip((bank.credit_score-650)/40+(bank.income_k-bank.loan_k)/30+bank.employment_years/15-1.2,-20,20)))
bank['approved']=rng.binomial(1,prob)
Xtr,Xte,ytr,yte=train_test_split(bank.drop(columns='approved'),bank.approved,test_size=.25,random_state=42,stratify=bank.approved)
m=GaussianNB().fit(Xtr,ytr); pred=m.predict(Xte); print('Accuracy:',accuracy_score(yte,pred)); print('Confusion matrix:\n',confusion_matrix(yte,pred)); print('Demo prediction (1=approved):',m.predict([[70,25,710,6]])[0])
```
