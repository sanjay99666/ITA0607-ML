```python
# Shared imports (available in standard Google Colab)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_classification, make_regression, make_blobs, fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, mean_squared_error, r2_score

np.random.seed(42)
print("Setup complete")
```

```python
from sklearn.ensemble import RandomForestRegressor
try:
 housing=fetch_california_housing()
 Xtr,Xte,ytr,yte=train_test_split(housing.data,housing.target,test_size=.2,random_state=42)
 m=RandomForestRegressor(n_estimators=100,max_depth=14,n_jobs=-1,random_state=42).fit(Xtr,ytr); p=m.predict(Xte)
 print('Target: median house value in units of $100,000'); print('RMSE:',round(mean_squared_error(yte,p)**.5,3),'R²:',round(r2_score(yte,p),3))
except Exception as e:
 print('California dataset unavailable (network/cache issue). Running included synthetic fallback.')
 X,y=make_regression(n_samples=700,n_features=8,noise=15,random_state=42); Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
 m=RandomForestRegressor(n_estimators=100,random_state=42).fit(Xtr,ytr); p=m.predict(Xte); print('Synthetic fallback RMSE:',mean_squared_error(yte,p)**.5,'R²:',r2_score(yte,p))
```
