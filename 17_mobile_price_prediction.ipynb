```python
# Shared imports (available in standard Google Colab)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_classification, make_regression, make_blobs, fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, mean_squared_error, r2_score

np.random.seed(42)
print("Setup complete")
```

```python
from sklearn.ensemble import RandomForestClassifier
rng=np.random.default_rng(17); n=800
phones=pd.DataFrame({'battery':rng.integers(1800,6000,n),'ram_mb':rng.choice([1024,2048,3072,4096,6144,8192],n),'storage_gb':rng.choice([16,32,64,128,256],n),'camera_mp':rng.integers(5,109,n),'screen_inches':rng.uniform(4.5,7.2,n)})
latent=phones.ram_mb/1800+phones.storage_gb/45+phones.camera_mp/35+phones.battery/5000+phones.screen_inches/7
phones['price_range']=pd.qcut(latent+ rng.normal(0,.3,n),4,labels=False)
Xtr,Xte,ytr,yte=train_test_split(phones.drop(columns='price_range'),phones.price_range,test_size=.25,random_state=42,stratify=phones.price_range)
m=RandomForestClassifier(n_estimators=120,random_state=42).fit(Xtr,ytr); print('Accuracy:',accuracy_score(yte,m.predict(Xte))); print('0=low, 3=very high; example:',m.predict([[4500,4096,128,48,6.4]])[0])
```
