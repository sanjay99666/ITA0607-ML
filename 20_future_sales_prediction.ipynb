```python
# Shared imports (available in standard Google Colab)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris, make_classification, make_regression, make_blobs, fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, mean_squared_error, r2_score

np.random.seed(42)
print("Setup complete")
```

```python
from sklearn.ensemble import RandomForestRegressor
rng=np.random.default_rng(20); months=pd.date_range('2018-01-01',periods=84,freq='MS')
t=np.arange(len(months)); sales=200+2.4*t+35*np.sin(2*np.pi*t/12)+rng.normal(0,12,len(t))
series=pd.DataFrame({'month':months,'sales':sales}); series['t']=t; series['month_num']=series.month.dt.month
series['sin_month']=np.sin(2*np.pi*series.month_num/12); series['cos_month']=np.cos(2*np.pi*series.month_num/12)
features=['t','month_num','sin_month','cos_month']; train=series.iloc[:-12]; test=series.iloc[-12:]
m=RandomForestRegressor(n_estimators=200,min_samples_leaf=2,random_state=42).fit(train[features],train.sales); pred=m.predict(test[features])
print('12-month test RMSE:',round(mean_squared_error(test.sales,pred)**.5,2),'R²:',round(r2_score(test.sales,pred),3))
plt.plot(train.month,train.sales,label='History'); plt.plot(test.month,test.sales,'o-',label='Actual'); plt.plot(test.month,pred,'o--',label='Forecast'); plt.legend(); plt.title('Monthly sales: last 12 months holdout'); plt.show()
```
